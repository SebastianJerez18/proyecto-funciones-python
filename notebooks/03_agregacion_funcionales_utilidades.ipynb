## 1. `sum()`
**Descripción:** suma todos los elementos numéricos de un iterable.

**Sintaxis:** `sum(iterable, start=0)`

**Parámetros:**
- `iterable`: colección de números (lista, tupla, rango, etc.).
- `start`: valor inicial al que se suma el resultado (por defecto, 0).

**Retorna:** la suma total (`int` o `float`).


```python
# Ejemplo 1: suma de enteros y de decimales
print(sum([1, 2, 3, 4]))
print(sum([0.5, 1.5, 2]))

# Ejemplo 2: uso de start y cálculo de un promedio
print(sum([10, 20, 30], 100))
notas = [4.0, 3.5, 5.0]
print(round(sum(notas) / len(notas), 2))
```

---
## 2. `min()`
**Descripción:** devuelve el menor elemento de un iterable o de varios argumentos.

**Sintaxis:** `min(iterable, key=None)` o `min(a, b, c, ...)`

**Parámetros:**
- `iterable`: colección de elementos comparables.
- `key`: función que define el criterio de comparación (opcional).

**Retorna:** el elemento más pequeño según el criterio.

```python
# Ejemplo 1: con argumentos sueltos y con una lista
print(min(4, 9, 2))
print(min([7, 3, 8]))

# Ejemplo 2: con cadenas, por orden alfabético y por longitud
palabras = ["mango", "kiwi", "uva"]
print(min(palabras))
print(min(palabras, key=len))
```

---
## 3. `max()`
**Descripción:** devuelve el mayor elemento de un iterable o de varios argumentos.

**Sintaxis:** `max(iterable, key=None)` o `max(a, b, c, ...)`

**Parámetros:**
- `iterable`: colección de elementos comparables.
- `key`: función que define el criterio de comparación (opcional).

**Retorna:** el elemento más grande según el criterio.

```python
# Ejemplo 1: con argumentos sueltos y con una lista
print(max(4, 9, 2))
print(max([7, 3, 8]))

# Ejemplo 2: con cadenas, por orden alfabético y por longitud
palabras = ["mango", "kiwi", "uva"]
print(max(palabras))
print(max(palabras, key=len))
```
