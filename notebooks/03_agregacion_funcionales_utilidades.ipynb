## 1. `sum()`
**Descripción:** suma todos los elementos numéricos de un iterable.

**Sintaxis:** `sum(iterable, start=0)`

**Parámetros:**
- `iterable`: colección de números (lista, tupla, rango, etc.).
- `start`: valor inicial al que se suma el resultado (por defecto, 0).

**Retorna:** la suma total (`int` o `float`).


```python
# Ejemplo 1: suma de enteros y de decimales
print(sum([1, 2, 3, 4]))
print(sum([0.5, 1.5, 2]))

# Ejemplo 2: uso de start y cálculo de un promedio
print(sum([10, 20, 30], 100))
notas = [4.0, 3.5, 5.0]
print(round(sum(notas) / len(notas), 2))
```

---
## 2. `min()`
**Descripción:** devuelve el menor elemento de un iterable o de varios argumentos.

**Sintaxis:** `min(iterable, key=None)` o `min(a, b, c, ...)`

**Parámetros:**
- `iterable`: colección de elementos comparables.
- `key`: función que define el criterio de comparación (opcional).

**Retorna:** el elemento más pequeño según el criterio.

```python
# Ejemplo 1: con argumentos sueltos y con una lista
print(min(4, 9, 2))
print(min([7, 3, 8]))

# Ejemplo 2: con cadenas, por orden alfabético y por longitud
palabras = ["mango", "kiwi", "uva"]
print(min(palabras))
print(min(palabras, key=len))
```

---
## 3. `max()`
**Descripción:** devuelve el mayor elemento de un iterable o de varios argumentos.

**Sintaxis:** `max(iterable, key=None)` o `max(a, b, c, ...)`

**Parámetros:**
- `iterable`: colección de elementos comparables.
- `key`: función que define el criterio de comparación (opcional).

**Retorna:** el elemento más grande según el criterio.

```python
# Ejemplo 1: con argumentos sueltos y con una lista
print(max(4, 9, 2))
print(max([7, 3, 8]))

# Ejemplo 2: con cadenas, por orden alfabético y por longitud
palabras = ["mango", "kiwi", "uva"]
print(max(palabras))
print(max(palabras, key=len))
```

---
## 4. `map()`
**Descripción:** aplica una función a cada elemento de uno o más iterables.

**Sintaxis:** `map(funcion, iterable, ...)`

**Parámetros:**
- `funcion`: función que se aplica a cada elemento.
- `iterable`: colección (o varias) cuyos elementos se transforman.

**Retorna:** un iterador (`map`). Para verlo como lista se usa `list()`.

```python
# Ejemplo 1: elevar al cuadrado cada número
numeros = [1, 2, 3, 4]
print(list(map(lambda x: x ** 2, numeros)))

# Ejemplo 2: convertir cadenas a enteros y combinar dos listas
print(list(map(int, ["10", "20", "30"])))
print(list(map(lambda a, b: a + b, [1, 2, 3], [10, 20, 30])))
```

## 5. `filter()`
**Descripción:** conserva solo los elementos de un iterable para los que una función devuelve `True`.

**Sintaxis:** `filter(funcion, iterable)`

**Parámetros:**
- `funcion`: función que devuelve `True` o `False` para cada elemento. Si es `None`, se descartan los valores falsos.
- `iterable`: colección a filtrar.

**Retorna:** un iterador (`filter`) con los elementos que cumplen la condición.

```python
# Ejemplo 1: quedarse con los números pares
print(list(filter(lambda x: x % 2 == 0, [1, 2, 3, 4, 5, 6])))

# Ejemplo 2: palabras largas y eliminación de valores vacíos
palabras = ["sol", "luna", "estrella", "mar"]
print(list(filter(lambda p: len(p) > 3, palabras)))
print(list(filter(None, [0, 1, "", "hola", None, 5])))
```

## 6. `any()`
**Descripción:** indica si **al menos un** elemento del iterable es verdadero.

**Sintaxis:** `any(iterable)`

**Parámetros:**
- `iterable`: colección de valores a evaluar.

**Retorna:** `True` si algún elemento es verdadero; `False` si todos son falsos o el iterable está vacío.

```python
# Ejemplo 1: listas con valores verdaderos y falsos
print(any([False, 0, 3]))
print(any([0, "", None]))
print(any([]))

# Ejemplo 2: con una condición (¿alguna nota es 4.0 o más?)
notas = [2.5, 4.0, 3.2]
print(any(n >= 4.0 for n in notas))
```

---
## 7. `all()`
**Descripción:** indica si **todos** los elementos del iterable son verdaderos.

**Sintaxis:** `all(iterable)`

**Parámetros:**
- `iterable`: colección de valores a evaluar.

**Retorna:** `True` si todos los elementos son verdaderos o si el iterable está vacío; `False` en caso contrario.

```python
# Ejemplo 1: listas con todos los valores verdaderos o no
print(all([True, 1, "a"]))
print(all([1, 0, 2]))
print(all([]))

# Ejemplo 2: con una condición (¿todos son mayores de edad?)
edades = [18, 21, 25]
print(all(e >= 18 for e in edades))
print(all(e >= 21 for e in edades))
```

## 8. `isinstance()`
**Descripción:** comprueba si un objeto es de un tipo (o de uno de varios tipos) determinado.

**Sintaxis:** `isinstance(objeto, tipo)`

**Parámetros:**
- `objeto`: valor a comprobar.
- `tipo`: clase, o tupla de clases, contra la que se compara.

**Retorna:** `True` si el objeto pertenece al tipo indicado; `False` en caso contrario.

```python
# Ejemplo 1: comprobaciones básicas
print(isinstance(5, int))
print(isinstance("hola", str))
print(isinstance(3.14, int))

# Ejemplo 2: varios tipos a la vez, y el caso de bool (que es subclase de int)
print(isinstance(10, (int, float)))
print(isinstance("10", (int, float)))
print(isinstance(True, int))
```

## 9. `open()`
**Descripción:** abre un archivo para leerlo o escribirlo. Se recomienda usarlo con `with`, que cierra el archivo automáticamente.

**Sintaxis:** `open(archivo, modo='r', encoding=None)`

**Parámetros:**
- `archivo`: nombre o ruta del archivo.
- `modo`: `'r'` lectura, `'w'` escritura (borra el contenido previo), `'a'` agregar al final.
- `encoding`: codificación del texto, por ejemplo `'utf-8'`.

**Retorna:** un objeto archivo con métodos como `read()`, `write()` y `readline()`.

```python
import os

# Ejemplo 1: escribir un archivo y leerlo completo
with open("ejemplo.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Hola\nSegunda línea\n")

with open("ejemplo.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()
print(contenido)  

# Ejemplo 2: agregar una línea y leer línea por línea
with open("ejemplo.txt", "a", encoding="utf-8") as archivo:
    archivo.write("Tercera línea\n")

with open("ejemplo.txt", "r", encoding="utf-8") as archivo:
    for numero, linea in enumerate(archivo, start=1):
        print(numero, linea.strip())

# Limpieza: se elimina el archivo de ejemplo para no dejarlo en el repositorio
os.remove("ejemplo.txt")
print("Archivo de ejemplo eliminado")
```

## 10. `help()`
**Descripción:** muestra la documentación de una función, clase o módulo. Es útil para consultar la sintaxis sin salir de Python.

**Sintaxis:** `help(objeto)`

**Parámetros:**
- `objeto`: función, método, clase o módulo del que se quiere ver la ayuda.

**Retorna:** `None`. Imprime la documentación en pantalla.

```python
# Ejemplo 1: ayuda de una función integrada
help(len)

# Ejemplo 2: ayuda de un método de las cadenas
help(str.upper)
```
