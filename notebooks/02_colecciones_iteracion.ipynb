Funciones documentadas: len, range, list, tuple, dict, set, sorted, enumerate, zip, reversed.

## 1. len()
*Descripción:* devuelve la cantidad de elementos de un objeto.

*Sintaxis:* len(objeto)

*Parámetros:*
- objeto: cadena, lista, tupla, diccionario, conjunto, etc.

*Retorna:* un entero con el número de elementos.

```python
# Ejemplo 1: longitud de una cadena y de una lista
print(len("Python"))
print(len([10, 20, 30, 40]))

# Ejemplo 2: longitud de un diccionario
alumno = {"nombre": "Ana", "edad": 20}
print(len(alumno))
```

## 2. range()
*Descripción:* genera una secuencia de números enteros, normalmente usada para repetir ciclos.

*Sintaxis:* range(stop) o range(start, stop, step)

*Parámetros:*
- start: valor inicial (por defecto, 0).
- stop: valor final, que *no* se incluye.
- step: incremento entre valores (por defecto, 1).

*Retorna:* un objeto range (secuencia inmutable de enteros).

```python
# Ejemplo 1: números del 0 al 4
for i in range(5):
    print(i)

# Ejemplo 2: pares del 2 al 10, y conversión a lista
print(list(range(2, 11, 2)))
```

## 3. list()
*Descripción:* crea una lista, o convierte un iterable (cadena, tupla, rango, etc.) en lista.

*Sintaxis:* list(iterable)

*Parámetros:*
- iterable: objeto recorrible. Si se omite, devuelve una lista vacía.

*Retorna:* una lista (list) con los elementos del iterable.

```python
# Ejemplo 1: de cadena y de tupla a lista
print(list("hola"))
print(list((1, 2, 3)))

# Ejemplo 2: lista vacía y lista a partir de range
print(list())
print(list(range(3)))
```

## 4. tuple()
*Descripción:* crea una tupla, o convierte un iterable en tupla. Las tuplas son inmutables: no se pueden modificar después de creadas.

*Sintaxis:* tuple(iterable)

*Parámetros:*
- iterable: objeto recorrible (lista, cadena, rango, etc.). Si se omite, devuelve una tupla vacía.

*Retorna:* una tupla (tuple) con los elementos del iterable.

```python
# Ejemplo 1: de lista y de cadena a tupla
print(tuple([1, 2, 3]))
print(tuple("abc"))

# Ejemplo 2: las tuplas son inmutables
coordenadas = (4, 7)
print(coordenadas[0])
try:
    coordenadas[0] = 10
except TypeError as error:
    print("Error:", error)
```

## 5. dict()
*Descripción:* crea un diccionario, que almacena pares clave: valor.

*Sintaxis:* dict(**kwargs) o dict(iterable_de_pares)

*Parámetros:*
- kwargs: pares nombre=valor.
- iterable_de_pares: una secuencia de pares (clave, valor).

*Retorna:* un diccionario (dict).

```python
# Ejemplo 1: crear con dict() a partir de pares y de nombres
print(dict([("a", 1), ("b", 2)]))
print(dict(nombre="Luis", edad=21))

# Ejemplo 2: acceder y agregar elementos
persona = {"nombre": "Luis", "edad": 21}
print(persona["nombre"])
persona["ciudad"] = "Bogotá"
print(persona)
```

## 6. set()
*Descripción:* crea un conjunto, una colección sin elementos repetidos y sin orden garantizado.

*Sintaxis:* set(iterable)

*Parámetros:*
- iterable: objeto recorrible. Si se omite, devuelve un conjunto vacío.

*Retorna:* un conjunto (set) sin duplicados.

```python
# Ejemplo 1: eliminar duplicados de una lista
print(set([1, 2, 2, 3, 3, 3]))

# Ejemplo 2: operaciones entre conjuntos
a = {1, 2, 3}
b = {3, 4, 5}
print("Unión:", a | b)
print("Intersección:", a & b)
print("Diferencia:", a - b)
```

## 7. sorted()
*Descripción:* devuelve una lista nueva con los elementos de un iterable ordenados. No modifica el original.

*Sintaxis:* sorted(iterable, key=None, reverse=False)

*Parámetros:*
- iterable: colección a ordenar.
- key: función que define el criterio de orden (opcional).
- reverse: si es True, ordena de mayor a menor.

*Retorna:* una lista (list) ordenada.

```python
# Ejemplo 1: orden ascendente y descendente
numeros = [5, 2, 9, 1]
print(sorted(numeros))
print(sorted(numeros, reverse=True))
print(numeros)  # la lista original no cambia

# Ejemplo 2: ordenar por longitud con key
palabras = ["pera", "fresa", "kiwi", "mango"]
print(sorted(palabras, key=len))
```

## 8. enumerate()
*Descripción:* recorre un iterable devolviendo pares (índice, elemento). Evita llevar un contador manual.

*Sintaxis:* enumerate(iterable, start=0)

*Parámetros:*
- iterable: colección a recorrer.
- start: valor inicial del índice (por defecto, 0).

*Retorna:* un objeto enumerate (iterador de tuplas).

```python
# Ejemplo 1: recorrer con índice
frutas = ["manzana", "pera", "uva"]
for i, fruta in enumerate(frutas):
    print(i, fruta)

# Ejemplo 2: empezar en 1 y convertir a lista
print(list(enumerate(frutas, start=1)))
```

## 9. zip()
*Descripción:* combina varios iterables elemento a elemento, formando tuplas. Se detiene cuando termina el más corto.

*Sintaxis:* zip(*iterables)

*Parámetros:*
- iterables: dos o más colecciones a combinar.

*Retorna:* un iterador de tuplas (zip).

```python
# Ejemplo 1: emparejar nombres con notas
nombres = ["Ana", "Luis", "Marta"]
notas = [4.5, 3.8, 4.9]
print(list(zip(nombres, notas)))
for nombre, nota in zip(nombres, notas):
    print(nombre, "->", nota)

# Ejemplo 2: longitudes distintas y creación de un diccionario
print(list(zip([1, 2, 3], ["a", "b"])))
print(dict(zip(["x", "y"], [10, 20])))
```

## 10. reversed()
*Descripción:* devuelve un iterador que recorre una secuencia en orden inverso. No modifica la secuencia original.

*Sintaxis:* reversed(secuencia)

*Parámetros:*
- secuencia: lista, tupla, cadena o rango (objeto con orden definido).

*Retorna:* un iterador (reversed). Para verlo como lista se usa list().

```python
# Ejemplo 1: invertir una lista y una cadena
print(list(reversed([1, 2, 3, 4])))
print("".join(reversed("python")))

# Ejemplo 2: contar hacia atrás
for i in reversed(range(1, 4)):
    print(i)
```
