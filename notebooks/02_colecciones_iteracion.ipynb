## 1. len()
*Descripción:* devuelve la cantidad de elementos de un objeto.

*Sintaxis:* len(objeto)

*Parámetros:*
- objeto: cadena, lista, tupla, diccionario, conjunto, etc.

*Retorna:* un entero con el número de elementos.

```python
# Ejemplo 1: longitud de una cadena y de una lista
print(len("Python"))
print(len([10, 20, 30, 40]))

# Ejemplo 2: longitud de un diccionario
alumno = {"nombre": "Ana", "edad": 20}
print(len(alumno))
```

## 2. range()
*Descripción:* genera una secuencia de números enteros, normalmente usada para repetir ciclos.

*Sintaxis:* range(stop) o range(start, stop, step)

*Parámetros:*
- start: valor inicial (por defecto, 0).
- stop: valor final, que *no* se incluye.
- step: incremento entre valores (por defecto, 1).

*Retorna:* un objeto range (secuencia inmutable de enteros).

```python
# Ejemplo 1: números del 0 al 4
for i in range(5):
    print(i)

# Ejemplo 2: pares del 2 al 10, y conversión a lista
print(list(range(2, 11, 2)))
```

## 3. list()
*Descripción:* crea una lista, o convierte un iterable (cadena, tupla, rango, etc.) en lista.

*Sintaxis:* list(iterable)

*Parámetros:*
- iterable: objeto recorrible. Si se omite, devuelve una lista vacía.

*Retorna:* una lista (list) con los elementos del iterable.

```python
# Ejemplo 1: de cadena y de tupla a lista
print(list("hola"))
print(list((1, 2, 3)))

# Ejemplo 2: lista vacía y lista a partir de range
print(list())
print(list(range(3)))
```
