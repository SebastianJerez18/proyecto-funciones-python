# Notebook 01: Entrada, conversión y funciones numéricas

Funciones documentadas: `print`, `input`, `type`, `int`, `float`, `str`, `bool`, `abs`, `round`, `pow`.

---
## 1. `print()`
**Descripción:** muestra uno o varios valores en pantalla.

**Sintaxis:** `print(*objetos, sep=' ', end='\n')`

**Parámetros:**
- `objetos`: los valores a mostrar.
- `sep`: separador entre valores (por defecto, un espacio).
- `end`: texto al final de la línea (por defecto, salto de línea).

**Retorna:** `None`.


```python
   # Ejemplo 1: imprimir varios valores
   print("Hola", "mundo", 2026)

   # Ejemplo 2: cambiar el separador y el final
   print("a", "b", "c", sep="-", end=" FIN\n")
```

   ---
   ## 2. `input()`
   **Descripción:** lee texto que el usuario escribe por teclado. Siempre devuelve un `str`.

   **Sintaxis:** `input(mensaje='')`

   **Parámetros:**
   - `mensaje`: texto que se muestra antes de esperar la entrada.

   **Retorna:** una cadena (`str`) con lo que escribió el usuario.

```python
   # Ejemplo 1: pedir un nombre
   nombre = input("¿Cómo te llamas? ")
   print("Hola,", nombre)

   # Ejemplo 2: convertir la entrada a número
   edad = int(input("¿Cuántos años tienes? "))
   print("El próximo año tendrás", edad + 1)
```

   ---
   ## 3. `type()`
   **Descripción:** devuelve el tipo de dato de un objeto.

   **Sintaxis:** `type(objeto)`

   **Parámetros:**
   - `objeto`: cualquier valor o variable.

   **Retorna:** la clase del objeto (por ejemplo, `int`, `str`, `list`).
   

```python
   # Ejemplo 1: tipos básicos
   print(type(10))
   print(type(3.14))
   print(type("texto"))

   # Ejemplo 2: tipos de colecciones
   print(type([1, 2, 3]))
   print(type({"a": 1}))
```

---
## 4. `int()`
**Descripción:** convierte un valor a número entero.

**Sintaxis:** `int(x, base=10)`

**Parámetros:**
- `x`: número o cadena a convertir.
- `base`: base numérica de la cadena (opcional).

**Retorna:** un entero (`int`). Trunca los decimales, no redondea.

```python
# Ejemplo 1: de decimal a entero (trunca)
print(int(7.9))

# Ejemplo 2: de cadena a entero, y de binario a entero
print(int("25") + 5)
print(int("1010", 2))
```

---
## 5. `float()`
**Descripción:** convierte un valor a número decimal.

**Sintaxis:** `float(x)`

**Parámetros:**
- `x`: número o cadena a convertir.

**Retorna:** un número de punto flotante (`float`).

```python
# Ejemplo 1: de entero a decimal
print(float(3))

# Ejemplo 2: de cadena a decimal
precio = float("19.99")
print(precio * 2)
```

---
## 6. `str()`
**Descripción:** convierte un valor a cadena de texto.

**Sintaxis:** `str(objeto)`

**Parámetros:**
- `objeto`: cualquier valor a convertir.

**Retorna:** una cadena (`str`).

```python
# Ejemplo 1: número a texto
edad = 20
print("Tengo " + str(edad) + " años")

# Ejemplo 2: lista a texto
print(str([1, 2, 3]))
print(type(str([1, 2, 3])))
```

---
## 7. `bool()`
**Descripción:** convierte un valor a booleano (`True` o `False`).

**Sintaxis:** `bool(x)`

**Parámetros:**
- `x`: valor a evaluar.

**Retorna:** `False` si el valor es vacío o cero (`0`, `""`, `[]`, `None`); `True` en los demás casos.

```python
# Ejemplo 1: valores que dan False
print(bool(0), bool(""), bool([]), bool(None))

# Ejemplo 2: valores que dan True
print(bool(5), bool("hola"), bool([0]))
```

---
## 8. `abs()`
**Descripción:** devuelve el valor absoluto de un número.

**Sintaxis:** `abs(x)`

**Parámetros:**
- `x`: número entero, decimal o complejo.

**Retorna:** el valor sin signo (o el módulo, si es complejo).

```python
# Ejemplo 1: enteros y decimales
print(abs(-15))
print(abs(-3.7))

# Ejemplo 2: distancia entre dos puntos de una recta
a, b = 4, 10
print("Distancia:", abs(a - b))
```

---
## 9. `round()`
**Descripción:** redondea un número a una cantidad de decimales.

**Sintaxis:** `round(numero, ndigitos=None)`

**Parámetros:**
- `numero`: valor a redondear.
- `ndigitos`: cantidad de decimales (si se omite, devuelve un entero).

**Retorna:** el número redondeado. Con valores en la mitad exacta (`2.5`), Python redondea al par más cercano.

```python
# Ejemplo 1: sin decimales
print(round(7.6))

# Ejemplo 2: con decimales, y caso de la mitad exacta
print(round(3.14159, 2))
print(round(2.5), round(3.5))
```

---
## 10. `pow()`
**Descripción:** eleva un número a una potencia.

**Sintaxis:** `pow(base, exp, mod=None)`

**Parámetros:**
- `base`: número base.
- `exp`: exponente.
- `mod`: si se indica, devuelve `(base ** exp) % mod`.

**Retorna:** el resultado de la potencia.

```python
# Ejemplo 1: potencias
print(pow(2, 5))
print(pow(9, 0.5))

# Ejemplo 2: con módulo
print(pow(3, 4, 5))
```
